```python
#cleaning RAM
ds['Ram']=ds['Ram'].str.replace('GB','').astype(int)
```

```python
#cleaning Weight
ds['Weight']=ds['Weight'].str.replace('kg','').astype(float)
```

```python
ds['Cpu_Brand'] = ds['Cpu'].apply(lambda x: x.split()[0]
```

```python
#droping cpu column because theres cpu_brand (cleaned) column
ds.drop('Cpu',axis=1,inplace=True)
```

```python
ds['Gpu'] = ds['Gpu'].apply(lambda x: x.split()[0])
```

```python
ds
```

```python
# Assistant
def get_hdd(memory):
    if 'HDD' in memory:
        if 'TB' in memory:
            # Extract the number before TB
            tb_value = memory.split('TB')[0].strip()
            if tb_value.replace('.', '', 1).isdigit():  # Handle decimal points
                return float(tb_value) * 1000  # Convert TB to GB equivalent
        elif 'GB' in memory:
            # Extract the number before GB
            gb_value = memory.split('GB')[0].strip()
            if gb_value.replace('.', '', 1).isdigit():  # Handle decimal points
                return float(gb_value)
        return 0  # Default if format is unrecognized
    return 0

def get_ssd(memory):
    if 'SSD' in memory:
        if 'TB' in memory:
            # Extract the number before TB
            tb_value = memory.split('TB')[0].strip()
            if tb_value.replace('.', '', 1).isdigit():  # Handle decimal points
                return float(tb_value) * 1000  # Convert TB to GB equivalent
        elif 'GB' in memory:
            # Extract the number before GB
            gb_value = memory.split('GB')[0].strip()
            if gb_value.replace('.', '', 1).isdigit():  # Handle decimal points
                return float(gb_value)
        return 0  # Default if format is unrecognized
    return 0

ds['HDD'] = ds['Memory'].apply(get_hdd)
ds['SSD'] = ds['Memory'].apply(get_ssd)

# Drop original memory column if no longer needed
ds.drop('Memory', axis=1, inplace=True)
```

```python
ds
```

```python
ds['OpSys'] = ds['OpSys'].replace({
    'Windows 10': 'Windows',
    'Windows 7': 'Windows',
    'Windows 8': 'Windows',
    'macOS': 'Mac',
    'Mac OS X': 'Mac',
    'Chrome OS': 'Others',
    'Linux': 'Others',
    'No OS': 'Others'
})
```

```python
from sklearn.preprocessing import LabelEncoder

label_enc_cols = ['Company', 'TypeName', 'Cpu_Brand', 'Gpu', 'OpSys']
le = LabelEncoder()

for col in label_enc_cols:
    ds[col] = le.fit_transform(ds[col])
```

```python
ds=ds[ds['Ram']>=8]
```

```python
# Extract Touchscreen
ds['Touchscreen'] = ds['ScreenResolution'].apply(lambda x: 1 if 'Touchscreen' in x else 0)

# Extract IPS Panel
ds['IPS'] = ds['ScreenResolution'].apply(lambda x: 1 if 'IPS' in x else 0)

# Extract X and Y resolution
resolution = ds['ScreenResolution'].str.extract(r'(\d+)x(\d+)')
ds['X_res'] = resolution[0].astype('int')
ds['Y_res'] = resolution[1].astype('int')

# Calculate PPI (Pixels Per Inch)
ds['PPI'] = ((ds['X_res']**2 + ds['Y_res']**2) ** 0.5) / ds['Inches']

# Drop original ScreenResolution
ds.drop('ScreenResolution', axis=1, inplace=True)
```
