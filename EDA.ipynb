```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
ds=pd.read_csv('myproject_data.csv')
```

```python
ds
```

```python
ds.drop(['Unnamed: 0', 'TypeName', 'Inches', 'X_res', 'Y_res', 'HDD'], axis=1, inplace=True)
```

```python
ds
```

```python

plt.figure(figsize=(12,6))
sns.heatmap(ds.corr(), annot=True, cmap='YlGnBu')
plt.title("Correlation Heatmap")
plt.show()
```

```python
#price vs Ram
sns.barplot(x='Ram', y='Price_euros', data=ds)
plt.title("Price vs RAM")
plt.ylabel("Price (Euros)")
plt.xlabel("RAM (GB)")
plt.show()
```

```python
#price vs brand
plt.figure(figsize=(12,6))
sns.barplot(x='Company', y='Price_euros', data=ds)
plt.title("Price vs Laptop Brand")
plt.xticks(rotation=45)
plt.show()
```

```python
#how ssd affects price
sns.scatterplot(x='SSD', y='Price_euros', data=ds)
plt.title("Price vs SSD Size")
plt.xlabel("SSD (GB)")
plt.ylabel("Price (Euros)")
plt.show()

```

```python
sns.boxplot(ds['Price_euros'])
plt.title("Laptop Price Distribution")
plt.show()
```
